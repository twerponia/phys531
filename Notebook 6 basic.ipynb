```python
import numpy as np
import matplotlib.pyplot as plt

def area(q, p):                                   # shoelace formula
    return 0.5*np.abs(np.dot(q, np.roll(p, -1)) - np.dot(p, np.roll(q, -1)))

# A blob of initial conditions: a circle in phase space
th = np.linspace(0, 2*np.pi, 200, endpoint=False)               
q0 = 1.0 + 0.3*np.cos(th)
p0 = 0.3*np.sin(th)
A0 = area(q0, p0)

h, N = 0.05, 400
t_final = N * h

# --- Hamiltonian: SHO with omega=1, H=(p^2+q^2)/2 ---
q, p = q0.copy(), p0.copy()
for i in range(N):
    q_new = q + h*p
    p_new = p - h*q_new                            
    q, p = q_new, p_new
A_ham = area(q, p)

# --- Non-Hamiltonian: damped oscillator ---
qd, pd = q0.copy(), p0.copy()
gamma = 0.1
for i in range(N):
    dq = pd
    dp = -qd - 2*gamma*pd
    qd, pd = qd + h*dq, pd + h*dp
A_damp = area(qd, pd)

print(f"A0 = {A0:.4f} | A_ham/A0 = {A_ham/A0:.4f} | A_damp/A0 = {A_damp/A0:.4f}")

# --- Plotting Phase Space Evolution ---
fig, axes = plt.subplots(1, 2, figsize=(12, 5), sharex=True, sharey=True)

# 1. Hamiltonian System Plot
axes[0].scatter(q0, p0, c='royalblue', s=12, alpha=0.5, label='$t=0$ (Initial)')
axes[0].scatter(q, p, c='crimson', s=12, alpha=0.8, label=f'$t={t_final:.1f}$ ($A/A_0={A_ham/A0:.3f}$)')
axes[0].set_title("Hamiltonian System (Area Preserving)")
axes[0].set_xlabel("$q$ (Position)")
axes[0].set_ylabel("$p$ (Momentum)")
axes[0].set_aspect('equal', adjustable='box')
axes[0].grid(True, linestyle=':', alpha=0.6)
axes[0].legend(loc='upper right')

# 2. Damped System Plot
axes[1].scatter(q0, p0, c='royalblue', s=12, alpha=0.5, label='$t=0$ (Initial)')
axes[1].scatter(qd, pd, c='darkorange', s=12, alpha=0.8, label=f'$t={t_final:.1f}$ ($A/A_0={A_damp/A0:.3f}$)')
axes[1].set_title("Damped System (Phase Space Collapse)")
axes[1].set_xlabel("$q$ (Position)")
axes[1].set_aspect('equal', adjustable='box')
axes[1].grid(True, linestyle=':', alpha=0.6)
axes[1].legend(loc='upper right')

plt.tight_layout()
plt.show()
```
