```python
import numpy as np
import matplotlib.pyplot as plt

def area(q, p):                                   # shoelace formula
    return 0.5*np.abs(np.dot(q, np.roll(p, -1)) - np.dot(p, np.roll(q, -1)))

# a blob of initial conditions: a circle in phase space
th = np.linspace(0, 2*np.pi, 200, endpoint = False)               
q0 = 1.0 + 0.3*np.cos(th); p0 = 0.3*np.sin(th)
A0 = area(q0, p0)

h, N = 0.05, 400
# --- Hamiltonian: SHO with omega=1, H=(p^2+q^2)/2 ---
q, p = q0.copy(), p0.copy()
for i in range(N):
    q_new = q + h*p
    p_new = p - h*q_new                            
    q, p = q_new, p_new
A_ham = area(q, p)

# --- non-Hamiltonian: damped oscillator ---
qd, pd = q0.copy(), p0.copy(); gamma = 0.1
for i in range(N):
    dq = pd; dp = -qd - 2*gamma*pd
    qd, pd = qd + h*dq, pd + h*dp

print("A0 =", A0, " A_ham/A0 =", A_ham/A0, " A_damp/A0 =", area(qd, pd)/A0)
```

```python
print("q range:", np.min(q), np.max(q))
print("qd range:", np.min(qd), np.max(qd))
```

```python
fig, axs = plt.subplots(1, 2, figsize=(10, 5))

# Plot t=0
axs[0].plot(q0, p0, 'b-', label='$t=0$')
axs[0].scatter(q0, p0, c='blue', s=10, alpha=0.5)
axs[0].plot(q, p, 'r-', label=f'$t={N*h}$')
axs[0].scatter(q, p, c='red', s=10, alpha=0.5)
axs[0].set_title('Hamiltonian (Symplectic)')
axs[0].set_xlabel('q')
axs[0].set_ylabel('p')
axs[0].legend()
axs[0].grid(True)
axs[0].axis('equal')

axs[1].plot(q0, p0, 'b-', label='$t=0$')
axs[1].scatter(q0, p0, c='blue', s=10, alpha=0.5)
axs[1].plot(qd, pd, 'g-', label=f'$t={N*h}$')
axs[1].scatter(qd, pd, c='green', s=10, alpha=0.5)
axs[1].set_title('Damped Oscillator')
axs[1].set_xlabel('q')
axs[1].set_ylabel('p')
axs[1].legend()
axs[1].grid(True)
axs[1].axis('equal')
plt.tight_layout()
```

```python
# Let's test if the current Hamiltonian system shears over larger N or if an anharmonic term is needed, or if we just plot as-is.
h, N_long = 0.05, 2000
q, p = q0.copy(), p0.copy()
for i in range(N_long):
    q_new = q + h*p
    p_new = p - h*q_new 
    q, p = q_new, p_new

qd, pd = q0.copy(), p0.copy(); gamma = 0.1
for i in range(N_long):
    dq = pd; dp = -qd - 2*gamma*pd
    qd, pd = qd + h*dq, pd + h*dp

print("Area initial:", A0)
print("Area ham/A0", area(q, p)/A0)
print("Area dam/A0", area(qd, pd)/A0)
```

```python
# Test Anharmonic potential: p_new = p - h * (q_new + 0.5 * q_new**3)
h, N_anh = 0.05, 800
q_a, p_a = q0.copy(), p0.copy()
for i in range(N_anh):
    q_a_new = q_a + h * p_a
    p_a_new = p_a - h * (q_a_new + 0.5 * q_a_new**3)
    q_a, p_a = q_a_new, p_a_new

fig, ax = plt.subplots(figsize=(6, 6))
ax.scatter(q0, p0, c='blue', s=5, label='$t=0$')
ax.scatter(q_a, p_a, c='red', s=5, label=f'$t={N_anh*h}$')
ax.plot(q_a, p_a, 'r-', alpha=0.5)
ax.set_aspect('equal')
ax.set_title("Anharmonic Oscillator (Shearing & Winding)")
ax.legend()
```

```python
# Pendulum: force = -sin(q)
h, N_pen = 0.05, 1200
th = np.linspace(0, 2*np.pi, 500, endpoint = False)               
q0_pen = 1.5 + 0.3*np.cos(th); p0_pen = 0.3*np.sin(th)
A0_pen = area(q0_pen, p0_pen)

q_p, p_p = q0_pen.copy(), p0_pen.copy()
for i in range(N_pen):
    q_p_new = q_p + h * p_p
    p_p_new = p_p - h * np.sin(q_p_new)
    q_p, p_p = q_p_new, p_p_new

print("Area initial pendulum:", A0_pen)
print("Area final pendulum:", area(q_p, p_p))

fig, ax = plt.subplots(figsize=(6, 6))
ax.scatter(q0_pen, p0_pen, c='blue', s=3, label='$t=0$')
ax.scatter(q_p, p_p, c='red', s=3, label=f'$t={N_pen*h}$')
ax.plot(q_p, p_p, 'r-', alpha=0.5)
ax.set_aspect('equal')
ax.set_title("Pendulum Oscillator (Shearing & Winding)")
ax.legend()
```
