```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

c = 1.0; m = 1.0; q = 1.0
B = np.array([0.0, 0.0, 1.0])

def gamma_from_p(p):
    return np.sqrt(1.0 + np.dot(p, p) / (m * c)**2)

def eom(t, y):
    r = y[0:3]
    p = y[3:6]
    v = p / (gamma_from_p(p) * m)                 
    drdt = v
    dpdt = q * np.cross(v, B)
    return np.concatenate([drdt, dpdt])

v0 = 0.9                               # initial speed, units of c
gamma0 = 1.0 / np.sqrt(1.0 - v0**2)
p0 = np.array([0.0, gamma0*v0 * m, 0.0])      # <-- look here
r0 = np.array([0.0, 0.0, 0.0])

T_cyclotron = 2 * np.pi * m * c / (q * 1.0)   # <-- look here

t_span = (0.0, 3 * T_cyclotron)
t_eval = np.linspace(*t_span, 3000)
sol = solve_ivp(eom, t_span, np.concatenate([r0, p0]),
                t_eval=t_eval, rtol=1e-10, atol=1e-12)

plt.plot(sol.y[0], sol.y[1])
plt.gca().set_aspect('equal')
plt.xlabel('x'); plt.ylabel('y'); plt.title('Relativistic cyclotron')
plt.show()
```
