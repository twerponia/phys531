```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt
```

```python
GM = 1.0

def kepler(t, state):
    x, y, vx, vy = state
    r = np.sqrt(x**2 + y**2)
    ax = -GM * x / r**3
    ay = -GM * y / r**3
    return [vx, vy, ax, ay]

# initial condition intended to give a bound ellipse
x0, y0   = 1.0, 0.0
vx0, vy0 = 0.0, 0.8
state0 = [x0, y0, vx0, vy0]

t_span = (0, 20)

sol = solve_ivp(kepler, t_span, state0, t_eval=np.linspace(0, 20, 1000), rtol=1e-9, atol=1e-9) 
```

```python
plt.plot(sol.y[0], sol.y[1])
plt.axis('equal')
plt.xlabel('x'); plt.ylabel('y')
plt.title('Kepler orbit')
plt.show()
```
