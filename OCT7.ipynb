```python
import numpy as np
g = np.diag([1.0, -1.0])          # Scheck's signature (+,-)

def boost_beta(b):
    gam = 1.0 / np.sqrt(1.0 - b*b)
    return np.array([[ gam,     -gam*b],
                     [-gam*b,    gam  ]])

L = boost_beta(0.6)
print("det      =", np.linalg.det(L))
print("residual =", np.max(np.abs(L.T @ g @ L - g)))
print("allclose =", np.allclose(L.T @ g @ L, g))
```

```python
def beta_from_gamma(gam):  return np.sqrt(1.0 - 1.0/gam**2)
def gamma_from_beta(b):    return 1.0 / np.sqrt(1.0 - b*b)

for name, gam in [("muon, g-2 ring",      2.93e1),
                  ("Tevatron proton",     1.044e3),
                  ("LHC proton, 7 TeV",   7.4605e3),
                  ("LEP electron",        2.0e5),
                  ("UHECR proton",        3.2e11)]:
    b = beta_from_gamma(gam)
    print(f"{name:20s} beta = {b:.17f}   gamma back = {gamma_from_beta(b):.6e}")
```

```python
for ex in range(0, 9):
    gam = 10.0**ex
    b   = beta_from_gamma(gam) if gam > 1 else 0.0
    L   = boost_beta(b)
    res = np.max(np.abs(L.T @ g @ L - g))
    print(f"gamma=1e{ex}   |residual| = {res:9.3e}   allclose = {np.allclose(L.T@g@L, g)}")
```

```python
def boost_eta(e):
    return np.array([[ np.cosh(e), -np.sinh(e)],
                     [-np.sinh(e),  np.cosh(e)]])

# (1) round-trip gamma -> eta -> gamma, far past where beta died
for ex in [1, 6, 8, 12, 50, 100, 200, 300]:
    gam = 10.0**ex
    eta = np.arccosh(gam)
    print(f"1e{ex:<4d}  eta = {eta:10.5f}   gamma back = {np.cosh(eta):.8e}")

# (2) composition: is the group law addition?
e1, e2 = 0.8, 1.3
P = boost_eta(e1) @ boost_eta(e2)
print("max|L1 L2 - L(e1+e2)| =", np.max(np.abs(P - boost_eta(e1+e2))))

# (3) recover 1-beta without ever forming beta
for ex in [8, 20, 50]:
    eta = np.arccosh(10.0**ex)
    print(f"gamma=1e{ex}:  1-beta = {2.0/(np.expm1(2*eta) + 2.0):.6e}")
```
